```python

fp = "/Users/isabelaconde/Desktop/data/NOAA_blended_winds_monthly/"

fps = sorted(glob.glob(os.path.join(fp, "*.nc")))

wnds = xr.open_mfdataset(
    fps,
    combine="by_coords"
)

wnds
```

```python
uas = wnds.u_wind.drop_vars('zlev').rename({'time': 'time', 'latitude': 'lat', 'longitude': 'lon'})
vas = wnds.v_wind.drop_vars('zlev').rename({'time': 'time', 'latitude': 'lat', 'longitude': 'lon'})
```

```python
uas_trend = find_trend_values(pacific_mask(uas)*uas).sel(lat = slice(-20,20), lon = slice(120,290))
vas_trend = find_trend_values(pacific_mask(vas)*vas).sel(lat = slice(-20,20), lon = slice(120,290))

# assign attributes
uas_trend = uas_trend.assign_attrs(uas.attrs)
uas_trend.attrs["comments"] = "Trend per month over 1982-2017 (monthly-mean zonal wind at 10 m)."
vas_trend = vas_trend.assign_attrs(vas.attrs)
vas_trend.attrs["comments"] = "Trend per month over 1982-2017 (monthly-mean meridional wind at 10 m)."

# save dt to NOAA blended winds directory
uas_trend.to_netcdf("/Users/isabelaconde/Desktop/data/NOAA_blended_winds_monthly/uas_noaa_blended_winds_trend_1982-2017.nc")
vas_trend.to_netcdf("/Users/isabelaconde/Desktop/data/NOAA_blended_winds_monthly/vas_noaa_blended_winds_trend_1982-2017.nc")
```
